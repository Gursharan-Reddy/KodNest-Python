```python
# 1. Convert Student Registration Values
# Question
# Title: Convert Student Registration Values

# A student registration system has received some values in data types that are not suitable for their intended use.

# Create the following variables:

# age_text with the string value "21"

# course_fee_text with the string value "499.50"

# attempts with the integer value 3

# Perform the following conversions:

# Convert age_text into an integer and store it in age.

# Convert course_fee_text into a float and store it in course_fee.

# Convert attempts into a string and store it in attempts_text.

# Display each converted value followed by its data type.

# Input
# The program does not require any input.

# Output
# Print the converted values and their data types exactly as shown:

# 21
# <class 'int'>
# 499.5
# <class 'float'>
# 3
# <class 'str'>

age_text = "21"
course_fee_text = "499.50"
attempts = 3

# Convert the values here
age = int(age_text)
print(age)
print(type(age))

course_fee = float(course_fee_text)
print(course_fee)
print(type(course_fee))

attempts_text = str(attempts)
print(attempts_text)
print(type(attempts_text))
```

```python
# 3. Read and Convert Student Details
# Question
# Read and Convert Student Details
# A student registration system receives the student’s name, age and course rating.

# Write a Python program that:

# Reads the student’s name as text.

# Reads the student’s age and converts it into an integer.

# Reads the course rating and converts it into a float.

# Displays the entered values.

# Displays the data type of each stored value.

# Input
# The program receives three values in the following order:

# Student name

# Student age

# Course rating

# Each value is provided on a new line.

# Output
# Display the student details and their data types exactly in the following format:

# Plaintext
# Student: Meera
# Age: 21
# Rating: 4.8
# <class 'str'>
# <class 'int'>
# <class 'float'>

a = input()
b = int(input())
c = float(input())

# Display the values
print(f"Student: {a}")
print(f"Age: {b}")
print(f"Rating: {c}")

# Display the data types
print(type(a))
print(type(b))
print(type(c))
```
