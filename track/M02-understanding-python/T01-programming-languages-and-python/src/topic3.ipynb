```python
# QuestionAnalyze a List of ScoresA trainer has recorded the assessment scores of several students
#  in a Python list.Write a Python program that:Reads the number of scores.
#  Reads each score and stores it in a list.Reads one score to search for.
#  Displays the highest score.Displays the lowest score.Displays the total of all scores.
#  Displays whether the searched score is present in the list.InputThe program receives input
#   in the following order:An integer $n$ representing the number of scores.
# n integers, with each score provided on a separate line.One integer representing the score to
# search for.OutputDisplay the results exactly in the following 
# format: Highest Score: 92
# Lowest Score: 61
# Total Score: 382
# Search Result: Found
# Constraints1 <= n <= 1000 <= score <= 100 
# Use an empty list to begin storing the scores.
# Use append() to add each score.Use max(), min() and sum().
# Use the in operator to search for the required score.Do not use input prompts.
# The output must exactly match the expected format.
# Sample Input
# 78
# 92
# 61
# 84
# 67
# 84
# Sample Output 
# Highest Score: 92
# Lowest Score: 61
# Total Score: 382
# Search Result: Found

# Read n and initialize the scores list
n = int(input())
scores = []

# Read and store all scores
for _ in range(n):
    val = int(input())
    scores.append(val)

# Read the score to search for
search_score = int(input())

# Calculate highest, lowest, and total using required built-in functions
highest = max(scores)
lowest = min(scores)
total = sum(scores)

# Display the highest, lowest and total scores
print(f"Highest Score: {highest}")
print(f"Lowest Score: {lowest}")
print(f"Total Score: {total}")

# Check search_score using the 'in' operator and display result
if search_score in scores:
    print("Search Result: Found")
else:
    print("Search Result: Not Found")
```

```python
# Question
# Create and Unpack a Student Record
# A student record contains a fixed set of values: the student’s name, course and score.

# Write a Python program that reads these three values, stores them in a tuple 
# and then unpacks the tuple into separate variables.

# Display the unpacked values in the required format.

# Input
# The program receives three inputs on separate lines:

# A string representing the student’s name

# A string representing the course name

# An integer representing the student’s score

# Output
# Display the unpacked values exactly in the following format:

# Name: Aarav
# Course: Python
# Score: 86

# Constraints
# Use input() to read the name and course.

# Use int(input()) to read the score.

# Sample Input

# Aarav
# Python
# 86
# Sample Output

# Name: Aarav
# Course: Python
# Score: 86

name = input()
course = input()
score = int(input())

# Create the tuple
student_record = (name, course, score)

# Unpack the tuple
student_name, student_course, student_score = student_record

# Display the unpacked values
print(f"Name: {student_name}")
print(f"Course: {student_course}")
print(f"Score: {student_score}")
```

```python
# Question
# Normalize and Process a Sentence
# Text entered by a user may contain unnecessary spaces, inconsistent letter casing and punctuation.

# Write a Python program that reads one complete sentence and processes it in the following order:

# Remove spaces from the beginning and end using strip().

# Convert the cleaned sentence to lowercase using lower().

# Remove all full stops (.) using replace().

# Split the normalized sentence into words using split().

# Join the words using a hyphen (-) as the separator.

# Convert the normalized sentence to uppercase using upper().

# Find the starting index of the word python using find().

# Display every processed result in the required format.

# Input
# The program receives one line containing a sentence.

# The sentence may contain:

# Leading or trailing spaces

# Uppercase and lowercase letters

# Multiple words

# Full stops

# Sample Input

#   Learn Python Step by Step.
# Sample Output

# Cleaned: Learn Python Step by Step.
# Normalized: learn python step by step
# Words: ['learn', 'python', 'step', 'by', 'step']
# Slug: learn-python-step-by-step
# Uppercase: LEARN PYTHON STEP BY STEP
# Python Position: 6

sentence = input()

# Clean and normalize the sentence
cleaned = sentence.strip()
print(f"Cleaned: {cleaned}")

normal = cleaned.lower().replace('.', '')
print(f"Normalized: {normal}")

# Split the sentence and create the slug
words = normal.split()
print(f"Words: {words}")

slug = '-'.join(words)
print(f"Slug: {slug}")

# Produce the uppercase form and search result
upper = normal.upper()
print(f"Uppercase: {upper}")

python_pos = normal.find('python')
print(f"Python Position: {python_pos}")
```

```python
# Question
# Demonstrate Shared-List Modification
# A student creates a list of three assessment scores.

# Another variable is assigned to the same list:

# alias_scores = original_scores
# This assignment does not create a new list. Both variables refer to the same list object.

# Write a Python program that:

# Reads three integer scores and stores them in original_scores.

# Creates alias_scores by assigning original_scores to it.

# Reads one replacement score and updates the first element through alias_scores.

# Reads one additional score and appends it through alias_scores.

# Displays both variables.

# Checks whether both variables refer to the same object using is.

# The output should demonstrate that changes made through the alias are visible through the original variable as well.

# Input
# The program receives five integers, each on a separate line:

# First original score

# Second original score

# Third original score

# Replacement score for index 0

# Additional score to append

# Output
# Display the final values using the following format:

# Original: [updated values]
# Alias: [updated values]
# Shared Object: True
# Constraints
# All input values are integers.

# Store the first three values in a list.

# Create the alias using direct assignment.

# Do not use .copy() or slicing while creating the alias.

# Update index 0 through alias_scores.

# Append the fifth input through alias_scores.

# Use the is operator to check object identity.

# Do not use input prompts.

original_scores = []

for _ in range(3):
    original_scores.append(int(input()))

alias_scores = original_scores

replacement_score = int(input())
additional_score = int(input())

# Modify the shared list through alias_scores
alias_scores[0] = replacement_score
alias_scores.append(additional_score)

# Display both variables and check whether they share one object
print(f"Original: {original_scores}")
print(f"Alias: {alias_scores}")
shared = alias_scores is original_scores
print(f"Shared Object: {shared}")
```

```python
# Question
# Process a Student Skill Record
# A student stores five technical skills in the order in which they plan to learn them.

# Write a Python program that:

# Reads five skill names and stores them in a list.

# Converts the list into a tuple named skill_record.

# Displays the complete tuple.

# Displays the first three skills.

# Displays the last two skills.

# Displays every alternate skill beginning with the first skill.

# Displays all skills in reverse order.

# Use tuple slicing to produce the required results.

# Input
# The program receives five skill names.

# Each skill is provided on a new line.

# Output
# Display the complete skill record and the sliced results in the following format:

# Skill Record: (...)
# First Three: (...)
# Last Two: (...)
# Alternate Skills: (...)
# Reversed Skills: (...)

skills = []

# Read and store five skills
for _ in range(5):
    val = input()
    skills.append(val)

# Convert the list into a tuple named skill_record
skill_record = tuple(skills)

# Create the required slices
first3 = skill_record[:3]
last2 = skill_record[-2:]
alt = skill_record[0::2]
rev = skill_record[::-1]

# Display all required results
print(f"Skill Record: {skill_record}")
print(f"First Three: {first3}")
print(f"Last Two: {last2}")
print(f"Alternate Skills: {alt}")
print(f"Reversed Skills: {rev}")
```
