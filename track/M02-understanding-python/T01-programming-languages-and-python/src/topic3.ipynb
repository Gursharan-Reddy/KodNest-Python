```python
# QuestionRead and Display List ElementsA list can store multiple values inside one variable.
# Write a Python program that reads the number of elements, accepts each integer one by one, 
# stores the values in a list and displays the completed list.The elements must remain 
# in the same order in which they are entered.InputThe first line contains an integer n,
#  representing the number of elements.The next $n$ lines each contain one integer to be 
#  stored in the list.
#  Output Display the complete list using Python list format.
#  For example:
#  [12, 25, 8, 19]
# Constraints: 1 <= n <= 100-1000 <= element <= 1000 
# Create an empty list before reading the elements.Use a loop to read exactly n values.
n = int(input())
numbers = []

# Read n integers and add them to the list
for _ in range(n):
    val = int(input())
    numbers.append(val)

print(numbers)
```

```python
# QuestionAnalyze a List of ScoresA trainer has recorded the assessment scores of several students
#  in a Python list.Write a Python program that:Reads the number of scores.
#  Reads each score and stores it in a list.Reads one score to search for.
#  Displays the highest score.Displays the lowest score.Displays the total of all scores.
#  Displays whether the searched score is present in the list.InputThe program receives input
#   in the following order:An integer $n$ representing the number of scores.
# n integers, with each score provided on a separate line.One integer representing the score to
# search for.OutputDisplay the results exactly in the following 
# format: Highest Score: 92
# Lowest Score: 61
# Total Score: 382
# Search Result: Found
# Constraints1 <= n <= 1000 <= score <= 100 
# Use an empty list to begin storing the scores.
# Use append() to add each score.Use max(), min() and sum().
# Use the in operator to search for the required score.Do not use input prompts.
# The output must exactly match the expected format.
# Sample Input
# 78
# 92
# 61
# 84
# 67
# 84
# Sample Output 
# Highest Score: 92
# Lowest Score: 61
# Total Score: 382
# Search Result: Found

# Read n and initialize the scores list
n = int(input())
scores = []

# Read and store all scores
for _ in range(n):
    val = int(input())
    scores.append(val)

# Read the score to search for
search_score = int(input())

# Calculate highest, lowest, and total using required built-in functions
highest = max(scores)
lowest = min(scores)
total = sum(scores)

# Display the highest, lowest and total scores
print(f"Highest Score: {highest}")
print(f"Lowest Score: {lowest}")
print(f"Total Score: {total}")

# Check search_score using the 'in' operator and display result
if search_score in scores:
    print("Search Result: Found")
else:
    print("Search Result: Not Found")
```

```python
# Question
# Create and Unpack a Student Record
# A student record contains a fixed set of values: the student’s name, course and score.

# Write a Python program that reads these three values, stores them in a tuple 
# and then unpacks the tuple into separate variables.

# Display the unpacked values in the required format.

# Input
# The program receives three inputs on separate lines:

# A string representing the student’s name

# A string representing the course name

# An integer representing the student’s score

# Output
# Display the unpacked values exactly in the following format:

# Name: Aarav
# Course: Python
# Score: 86

# Constraints
# Use input() to read the name and course.

# Use int(input()) to read the score.

# Sample Input

# Aarav
# Python
# 86
# Sample Output

# Name: Aarav
# Course: Python
# Score: 86

name = input()
course = input()
score = int(input())

# Create the tuple
student_record = (name, course, score)

# Unpack the tuple
student_name, student_course, student_score = student_record

# Display the unpacked values
print(f"Name: {student_name}")
print(f"Course: {student_course}")
print(f"Score: {student_score}")
```

```python
# Question
# Normalize and Process a Sentence
# Text entered by a user may contain unnecessary spaces, inconsistent letter casing and punctuation.

# Write a Python program that reads one complete sentence and processes it in the following order:

# Remove spaces from the beginning and end using strip().

# Convert the cleaned sentence to lowercase using lower().

# Remove all full stops (.) using replace().

# Split the normalized sentence into words using split().

# Join the words using a hyphen (-) as the separator.

# Convert the normalized sentence to uppercase using upper().

# Find the starting index of the word python using find().

# Display every processed result in the required format.

# Input
# The program receives one line containing a sentence.

# The sentence may contain:

# Leading or trailing spaces

# Uppercase and lowercase letters

# Multiple words

# Full stops

# Sample Input

#   Learn Python Step by Step.
# Sample Output

# Cleaned: Learn Python Step by Step.
# Normalized: learn python step by step
# Words: ['learn', 'python', 'step', 'by', 'step']
# Slug: learn-python-step-by-step
# Uppercase: LEARN PYTHON STEP BY STEP
# Python Position: 6

sentence = input()

# Clean and normalize the sentence
cleaned = sentence.strip()
print(f"Cleaned: {cleaned}")

normal = cleaned.lower().replace('.', '')
print(f"Normalized: {normal}")

# Split the sentence and create the slug
words = normal.split()
print(f"Words: {words}")

slug = '-'.join(words)
print(f"Slug: {slug}")

# Produce the uppercase form and search result
upper = normal.upper()
print(f"Uppercase: {upper}")

python_pos = normal.find('python')
print(f"Python Position: {python_pos}")
```
