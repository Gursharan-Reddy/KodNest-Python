```python
# Question
# Build a Simple Calculator Function
# A calculator receives two numbers and an operator.

# Write a Python program that defines a function named calculate().

# The function must accept two numbers and one operator, perform the required calculation and return the result.

# The supported operators are:

# + for addition

# - for subtraction

# * for multiplication

# / for division

# Call the function and print the returned result.

# Input
# The first line contains the first integer.

# The second line contains the second integer.

# The third line contains the operator.

# Output
# Print the result of the calculation.  

def calculate(first_number, second_number, operator):
    # Write your code here
    if operator == '+':
        return first_number + second_number
    elif operator == '-':
        return first_number - second_number
    elif operator == '*':
        return first_number * second_number
    elif operator == '/':
        return first_number / second_number


first_number = int(input())
second_number = int(input())
operator = input().strip()

result = calculate(first_number, second_number, operator)
print(result)
```

```python
# QuestionConvert a Number Sign Checker into a FunctionYou previously created a program 
# that checks whether a number is positive, negative or zero.Rewrite that logic as a reusable 
# function named check_sign.The function must receive one number and 
# return:Positive when the number is greater than zeroNegative when the number is less than zero
# Zero when the number is zeroRead one integer, call the function and 
# print the returned value.InputA single integer.OutputPrint Positive, Negative or Zero.

def check_sign(number):
    # Write your code here
    if number > 0:
        return "Positive"
    elif number == 0:
        return "Zero"
    else:
        return "Negative"


number = int(input())
result = check_sign(number)
print(result)
```
