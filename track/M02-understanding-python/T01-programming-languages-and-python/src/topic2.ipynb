```python
# Question
# Check Positive, Negative or Zero
# A number can be positive, negative or zero.

# Write a Python program that reads one integer and identifies its category.

# Use if, elif and else so that exactly one result is displayed.

# Input
# The program receives one integer.

# Output
# Display the result in one of the following formats:

# Plaintext
# Number is Positive
# Number is Negative
# Number is Zero
# Constraints
# -100000 <= number <= 100000

# Use int(input()) to read the value.

# Use if, elif and else.

# Do not use multiple separate if statements.

# Do not add an input prompt.

# Display the output exactly as shown.

# Read the number
number = int(input())

# Check whether the number is positive, negative or zero
if number > 0:
    print("Number is Positive")
elif number == 0:
    print("Number is Zero")
else:
    print("Number is Negative")
```

```python
# Question
# Student Eligibility Checker
# A student can attend the placement preparation session only when the required academic and project criteria are satisfied.

# Write a Python program that reads the student's marks, attendance percentage and project completion status.

# The student is eligible only when:

# Marks are greater than or equal to 60.

# Attendance is greater than or equal to 75.

# Project completion status is yes.

# First, use a compound condition to check the marks and attendance. Then, use a nested condition to check whether the project is completed.

# Input
# The program receives three inputs:

# An integer representing the marks.

# An integer representing the attendance percentage.

# A string representing the project completion status: yes or no.

# Each input is provided on a new line.

# Output
# Display:
# Eligible
# or
# Not Eligible

# Read marks, attendance and project completion status
marks = int(input())
attendance = int(input())
Status = input()

# Check the academic requirements
if marks >= 60 and attendance >= 75:
    if Status == "yes":
        print("Eligible")
    else:
        print("Not Eligible")
else:
    print("Not Eligible")
```

```python
# Question
# Calculate Total Using a while Loop
# Write a Python program that reads a positive integer n and calculates the sum of all integers from 1 to n using a while loop.

# For example, if n is 5, the program must calculate:
# 1 + 2 + 3 + 4 + 5 = 15

# Input
# The program receives one positive integer n.

# Output
# Display the calculated total in the following format:

# Plaintext
# Total: 15
# Constraints
# 1 <= n <= 1000

# Use int(input()) to read the value.

# Use a while loop.

# Initialize a counter variable with 1.

# Initialize a total variable with 0.

# Add each counter value to the total.

# Update the counter during every iteration.

# Do not use sum().

# Do not use a for loop.

# Read the value of n
n = int(input())

# Initialize the counter and total
count = 1
total = 0

# Calculate the total using a while loop
while count <= n:
    total += count
    count += 1

# Display the total
print(f"Total: {total}")
```

```python
# Question 
# Display a Number Sequence and Word CharactersWrite a Python program that reads a positive integer N and a word.
# First, use a for loop with range() to print all numbers from 1 to N.
# Then, use direct string iteration to print every character of the word.
# Print each number and each character on a separate line under the appropriate heading.
# InputThe first line contains a positive integer $N$.The second line contains a word.
# Output
# First,
# print:PlaintextNumbers:
# Then print the numbers from 1 to N, with each number on a new line.
# After that, 
# print:PlaintextCharacters:
# Then print every character of the word on a new line.
# Constraints1 <= N <= 100The word contains only English letters.
# The word contains at least one character.Use range(1, N + 1) to generate the numbers.
# Iterate directly over the word to access its characters.Do not use indexing to read the characters.
# Do not add input prompts.Sample InputPlaintext3
# Code
# Sample Output 
# Numbers:
# 1
# 2
# 3
# Characters:
# C
# o
# d
# e

# Read the number and word
num = int(input())
char = input()

# Print the number sequence
print("Numbers:")
for i in range(1, num + 1):
    print(i)

# Print the characters
print("Characters:")
for j in char:
    print(j)
```

```python
# Question: Analyze Multiples of ThreeWrite a Python program that reads a limit and a target number.
# Examine every number from 1 to the given limit.For numbers that are divisible by 3:
# Count how many such numbers are present.Calculate their total.Check whether the target number 
# is one of them.Display the count, total, and target-search result in the required format.
# Input The first line contains an integer limit.
# The second line contains an integer target.
# Output Display the number of multiples of 3 in the following format: Count: 3
# Display their total in the following format: Sum: 18
# Display whether the target was found among the multiples of 3: Target Found: Yes
# or Target Found:
# for loop with range().Examine numbers from 1 through limit.Use % to identify multiples of 3.
# Initialize the counter to 0.Initialize the accumulator to 0.Initialize the search flag to False.
# Do not use a mathematical formula to calculate the count or total.Do not use lists, sets or other 
# collections.Do not add input prompts.Display the output exactly as shown.
# Sample Test CaseSample Input: 10
# 6
# Sample Output:: 3
# Sum: 18
# Target Found: Yes

limit = int(input())
target = int(input())

count = 0
total = 0
found = False

# Examine every number from 1 to the limit
for i in range(1, limit + 1):
    if i % 3 == 0:
        count += 1
        total += i
        if i == target:
            found = True

# Display the count, total and search result
print(f"Count: {count}")
print(f"Sum: {total}")

if found:
    print("Target Found: Yes")
else:
    print("Target Found: No")
```

```python
# Read how many numbers will be entered
number_count = int(input())

# Initialize the counters and total
positive_count = 0
negative_count = 0
zero_count = 0
total = 0

# Read and analyze each number
for _ in range(number_count):
    val = int(input())
    
    if val > 0:
        positive_count += 1
    elif val < 0:
        negative_count += 1
    else:
        zero_count += 1
        
    total += val

# Display the final analysis
print(f"Positive Count: {positive_count}")
print(f"Negative Count: {negative_count}")
print(f"Zero Count: {zero_count}")
print(f"Total: {total}")
```
